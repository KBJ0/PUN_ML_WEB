```python
import sys, os, subprocess, pathlib, time
from dotenv import load_dotenv

WEB_ROOT = r"C:\Users\gatar\PythonProject\web"
ENV_PATH = pathlib.Path(WEB_ROOT, "app", ".env")   # ✅ app/.env 에서 읽기
load_dotenv(ENV_PATH, override=True)               # ✅ .env 로드

env = os.environ.copy()                            # ✅ 로드 후에 복사
port = "8001"                                      # 이미 8000 쓰면 충돌나니 8001

# uvicorn 서브프로세스 시작
proc = subprocess.Popen(
    [sys.executable, "-m", "uvicorn", "app.main:app",
     "--host", "127.0.0.1", "--port", port, "--reload", "--no-access-log"],
    cwd=WEB_ROOT,                # ✅ 작업 디렉터리는 web
    env=env,                     # ✅ .env가 반영된 환경 전달
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
)

# 첫 로그 몇 줄 보여주기
time.sleep(1.5)
for _ in range(10):
    line = proc.stdout.readline()
    if not line:
        break
    print(line.rstrip())

print(f"\n접속: http://127.0.0.1:{port}  (멈추려면 다음 셀에서 proc.terminate())")
```

```python
import os, smtplib
from email.message import EmailMessage
from dotenv import load_dotenv

load_dotenv("C:/Users/gatar/PythonProject/web/app/.env")

msg = EmailMessage()
msg["Subject"] = "테스트 메일"
msg["From"] = os.getenv("EMAIL_FROM")
msg["To"] = os.getenv("EMAIL_FROM")
msg.set_content("SMTP 설정 테스트 중입니다.")

with smtplib.SMTP_SSL(os.getenv("SMTP_HOST"), int(os.getenv("SMTP_PORT"))) as s:
    s.login(os.getenv("SMTP_USERNAME"), os.getenv("SMTP_PASSWORD"))
    s.send_message(msg)

print("메일 발송 완료")
```

```python
import os, sqlalchemy as sa, traceback, pathlib
WEB_ROOT = r"C:\Users\gatar\PythonProject\web"
from dotenv import load_dotenv; load_dotenv(pathlib.Path(WEB_ROOT, ".env"))

url = os.getenv("DATABASE_URL")
print("DATABASE_URL:", url)

try:
    eng = sa.create_engine(url, pool_pre_ping=True, pool_recycle=3600, echo=False)
    with eng.connect() as conn:
        print("VERSION:", conn.exec_driver_sql("SELECT VERSION()").scalar())
        print("SELECT 1:", conn.exec_driver_sql("SELECT 1").scalar())
    print("✅ DB 연결 성공")
except Exception as e:
    print("❌ DB 연결 실패:", type(e).__name__, e)
    traceback.print_exc(limit=5)
```

```python
from app.db import Base, engine           # ← 이렇게

from fastapi import FastAPI, Request
from fastapi.responses import HTMLResponse
from fastapi.staticfiles import StaticFiles
from fastapi.templating import Jinja2Templates
from app.routers import todos as todos_router

app = FastAPI(title="Todo REST API")

# Static & Templates
app.mount("/static", StaticFiles(directory="app/static"), name="static")
templates = Jinja2Templates(directory="app/templates")

# DB 초기화
@app.on_event("startup")
def on_startup():
    Base.metadata.create_all(bind=engine)

# REST 라우터
app.include_router(todos_router.router)

# 페이지
@app.get("/", response_class=HTMLResponse)
def index(request: Request):
    return templates.TemplateResponse("index.html", {"request": request, "title": "Todo List"})
```
